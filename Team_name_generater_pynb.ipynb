# 🚀 Modern & Creative Team Name Generator
This interactive tool generates professional and creative team names tailored to your industry, aesthetic preference, and keywords.

```python
import random
import ipywidgets as widgets
from IPython.display import display, HTML

# Defining styling for the generator
style_html = """
<style>
    .container-box {
        background: linear-gradient(135deg, #1e1e2f 0%, #252540 100%);
        padding: 30px;
        border-radius: 15px;
        box-shadow: 0 10px 25px rgba(0,0,0,0.3);
        font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
        color: #ffffff;
        max-width: 650px;
        margin: auto;
    }
    .header-title {
        text-align: center;
        background: linear-gradient(45deg, #ff7b00, #ffae00, #00d2ff);
        -webkit-background-clip: text;
        -webkit-text-fill-color: transparent;
        font-size: 28px;
        font-weight: 800;
        margin-bottom: 10px;
    }
    .subtitle {
        text-align: center;
        color: #a0a0b0;
        font-size: 14px;
        margin-bottom: 25px;
    }
    .result-card {
        background: rgba(255, 255, 255, 0.07);
        border-left: 5px solid #00d2ff;
        padding: 15px;
        border-radius: 8px;
        margin-top: 20px;
        animation: fadeIn 0.5s ease-in-out;
    }
    .result-title {
        font-size: 14px;
        color: #00d2ff;
        text-transform: uppercase;
        letter-spacing: 1.5px;
        margin-bottom: 5px;
    }
    .result-name {
        font-size: 22px;
        font-weight: bold;
        color: #ffffff;
    }
    @keyframes fadeIn {
        from { opacity: 0; transform: translateY(10px); }
        to { opacity: 1; transform: translateY(0); }
    }
</style>
"""
display(HTML(style_html))

# Word banks mapped to industries and vibes
adjectives = {
    "Creative": ["Apex", "Quantum", "Vivid", "Cognitive", "Synapse", "Prismatic", "Nexus", "Elysian", "Catalyst", "Aether"],
    "Tech": ["Cyber", "Binary", "Vector", "Algorithmic", "Cloud", "Helix", "Silicon", "Neural", "Matrix", "Syntax"],
    "Business / Professional": ["Summit", "Vanguard", "Synergy", "Pinnacle", "Alliance", "Crest", "Legacy", "Beacon", "Charter", "Apex"],
    "Bold & Fierce": ["Rogue", "Maverick", "Titan", "Vortex", "Blaze", "Phoenix", "Phantom", "Apex", "Velocity", "Obsidian"]
}

nouns = {
    "Creative": ["Guild", "Lab", "Studio", "Collective", "Forge", "Minds", "Wave", "Engine", "Canvas", "Innovators"],
    "Tech": ["Systems", "Nodes", "Bots", "Labs", "Devs", "Network", "Logic", "Matrix", "Architects", "Coders"],
    "Business / Professional": ["Consulting", "Partners", "Group", "Holdings", "Ventures", "Capital", "Advisors", "Strategies", "Syndicate"],
    "Bold & Fierce": ["Warriors", "Squad", "Force", "Knights", "Wolves", "Raiders", "Outlaws", "Brigade", "Rangers"]
}

# Create interactive widgets
industry_dropdown = widgets.Dropdown(
    options=["Tech", "Creative", "Business / Professional", "Bold & Fierce"],
    value="Tech",
    description="Vibe/Industry:"
)

keyword_input = widgets.Text(
    value="",
    placeholder="e.g., Pixel, Alpha (Optional)",
    description="Keyword:"
)

generate_button = widgets.Button(
    description="✨ Generate Team Names",
    button_style="info",
    layout=widgets.Layout(width='100%', height='45px', margin='20px 0px 0px 0px')
)

output_area = widgets.Output()

# Widget layout container
form_container = widgets.VBox([
    industry_dropdown,
    keyword_input,
    generate_button,
    output_area
], layout=widgets.Layout(padding='10px', width='100%'))

# Click action handler
def on_button_clicked(b):
    with output_area:
        output_area.clear_output()
        vibe = industry_dropdown.value
        kw = keyword_input.value.strip()

        # Generate a list of suggestions
        suggestions = []
        for _ in range(3):
            adj = random.choice(adjectives[vibe])
            noun = random.choice(nouns[vibe])

            if kw:
                pattern = random.choice([
                    f"{kw} {noun}",
                    f"{adj} {kw}",
                    f"The {adj} {kw} {noun}"
                ])
            else:
                pattern = f"{adj} {noun}"

            suggestions.append(pattern)

        # Render UI result block
        results_html = "<div class='result-card'>"
        results_html += "<div class='result-title'>💡 Generated Team Concepts:</div>"
        for idx, name in enumerate(suggestions, 1):
            results_html += f"<div class='result-name' style='margin-bottom:12px;'>{idx}. {name}</div>"
        results_html += "</div>"

        display(HTML(results_html))

generate_button.on_click(on_button_clicked)

# Render full styled container inside the notebook
display(HTML("""
<div class='container-box'>
    <div class='header-title'>⚡ Team Name Forge</div>
    <div class='subtitle'>Select a vibe, add custom keywords, and generate your team identity.</div>
"""))
display(form_container)
display(HTML("</div>"))
```
